# STEP 4 — CAMPUS WALLET CALCULATION

Setelah memiliki **Campus Wallet Account**, sistem harus menghitung kondisi saldo mahasiswa berdasarkan transaksi yang terjadi.

Mahasiswa memiliki tiga informasi keuangan:

* **Initial Balance** → saldo sebelum transaksi
* **Incoming Funds** → dana yang masuk ke akun
* **Outgoing Funds** → dana yang digunakan atau keluar dari akun

Hitung **Final Balance** berdasarkan transaksi tersebut.

Gunakan proses perhitungan secara **bertahap**, bukan langsung menggunakan satu rumus.

### Ketentuan

1. Buat variabel `final_balance` dan isi dengan nilai `saldo_awal`.
2. Tambahkan `dana_masuk` ke `final_balance` menggunakan assignment operator `+=`.
3. Kurangi `dana_keluar` dari `final_balance` menggunakan assignment operator `-=`.
4. Tampilkan informasi akun dan hasil transaksi.
5. Gunakan minimal **dua assignment operators**, yaitu `+=` dan `-=`.

### Expected Result

Gunakan contoh data berikut untuk memastikan program berjalan dengan benar:

```text
Initial Balance : 250000
Incoming Funds  : 500000
Outgoing Funds  : 175000
```

Proses yang diharapkan:

```text
Final Balance = 250000
Final Balance = 250000 + 500000
Final Balance = 750000
Final Balance = 750000 - 175000
Final Balance = 575000
```

Sehingga hasil akhirnya:

```text
Wallet Summary
----------------------------------------
Initial Balance : 250000
Incoming Funds  : 500000
Outgoing Funds  : 175000
Final Balance   : 575000
Account No.     : 1001
```

**Catatan:** `Account No.` harus berasal dari hasil STEP 3 menggunakan NRP mahasiswa masing-masing.


````python
# ================================================
# STEP 4 - CAMPUS WALLET CALCULATION
# ================================================

# Start Final Balance from Initial Balance
final_balance = saldoAwal

# Add Incoming Funds
print(final_balance)
final_balance += danaMasuk
print(final_balance)

# Subtract Outgoing Funds
print(final_balance)
final_balance -= danaKeluar
print(final_balance)

# Display Wallet Summary
print("Wallet Summary")
print("----------------------------------------")
print("Initial Balance :",saldoAwal)
print("Incoming Funds  :",danaMasuk)
print("Outgoing Funds  :",danaKeluar)
print("Final Balance   :",final_balance)
print("Account No.     :",account_number)
````

# STEP 5 — COMPARISON CHECK

Pada tahap ini, sistem akan memeriksa kondisi **Campus Wallet** menggunakan **comparison operators**.

Buat tiga variabel Boolean berdasarkan kondisi berikut:

### 1. Balance Check

Periksa apakah `final_balance` **lebih besar atau sama dengan Rp500.000**.

Gunakan operator:

```text
>=
```

Simpan hasilnya ke dalam variabel:

```text
balance_check
```

### 2. Transaction Check

Periksa apakah `dana_keluar` **lebih kecil atau sama dengan `dana_masuk`**.

Gunakan operator:

```text
<=
```

Simpan hasilnya ke dalam variabel:

```text
outgoing_check
```

### 3. Account Check

Periksa apakah `account_number` **lebih besar dari 0**.

Gunakan operator:

```text
>
```

Simpan hasilnya ke dalam variabel:

```text
account_check
```

### Expected Result

Jika data yang digunakan:

```text
Final Balance  = 575000
Incoming Funds = 500000
Outgoing Funds = 175000
Account Number = 1001
```

Maka hasilnya:

```text
Comparison Check
----------------------------------------
Balance >= 500000    : True
Outgoing <= Incoming : True
Account No. > 0      : True
```

**Catatan:** Semua hasil pemeriksaan harus berupa nilai Boolean, yaitu `True` atau `False`.


````python
# ================================================
# STEP 5 - COMPARISON CHECK
# ================================================
balance_check = final_balance >= 500000
outgoing_check = danaKeluar <= danaMasuk
account_check = account_number > 0

print("Comparison Check")
print("----------------------------------------")
# ================jawab disini================
print("Balance >= 500000    :",balance_check >= 500000)
print("Outgoing <= Incoming :",danaKeluar <= danaMasuk)
print("Account No. > 0      :",account_number > 0)
````

# STEP 6 — LOGICAL CHECK

Pada tahap ini, sistem akan menggabungkan beberapa hasil pemeriksaan dari STEP 5 menggunakan **logical operators**.

Gunakan tiga logical operators berikut:

* `and` → kedua kondisi harus benar
* `or` → salah satu kondisi benar
* `not` → membalik hasil Boolean

### 1. Wallet Ready

Campus Wallet dinyatakan **Ready** jika:

* `balance_check` bernilai `True`, **dan**
* `outgoing_check` bernilai `True`

Gunakan operator `and`.

Simpan hasilnya ke dalam variabel:

```text
wallet_ready
```

---

### 2. Account Status

Akun dapat digunakan jika:

* `account_check` bernilai `True`, **atau**
* `balance_check` bernilai `True`

Gunakan operator `or`.

Simpan hasilnya ke dalam variabel:

```text
account_usable
```

---

### 3. Wallet Not Ready

Buat pemeriksaan kebalikan dari `wallet_ready`.

Gunakan operator `not`.

Simpan hasilnya ke dalam variabel:

```text
not_wallet_ready
```

---

### Expected Result

Jika hasil dari STEP 5 adalah:

```text
balance_check  = True
outgoing_check = True
account_check  = True
```

Maka hasil STEP 6 adalah:

```text
Logical Check
----------------------------------------
Wallet Ready     : True
Account Usable   : True
Not Wallet Ready : False
```

**Catatan:** Semua hasil harus berupa Boolean (`True` atau `False`).


````python
# ================================================
# STEP 6 - LOGICAL CHECK
# ================================================

# Wallet is ready if both conditions are True
wallet_ready = final_balance >= 500000 and danaKeluar <= danaMasuk

# Account can be used if at least one condition is True
account_usable = final_balance >= 500000 or account_number > 0

# Wallet is not ready if both conditions are True
wallet_not_ready = not wallet_ready

# Reverse the wallet_ready result
wallet_ready = not wallet_not_ready


# Display the results
print("\nLogical Check")
print("----------------------------------------")
# ================jawab disini================
print("Wallet Ready     :",wallet_ready)
print("Account Usable   :",account_usable)
print("Not Wallet Ready :",wallet_not_ready)

````

Campus Wallet digunakan untuk mencatat aktivitas mahasiswa di lingkungan kampus.

Pada tahap ini, sistem akan memeriksa apakah **aktivitas yang dimasukkan mahasiswa** termasuk dalam aktivitas yang tercatat di sistem.

Gunakan aktivitas berikut sebagai data aktivitas yang tersedia:

* `KKN`
* `Magang`

Aktivitas tersebut merupakan contoh kegiatan mahasiswa yang dapat dicatat dalam sistem Campus Wallet.

### Tugas

1. Buat variabel `activities` yang berisi seluruh aktivitas yang tersedia.
2. Gunakan operator **`in`** untuk memeriksa apakah `activity` yang dimasukkan mahasiswa terdapat dalam daftar aktivitas.
3. Simpan hasilnya ke dalam variabel `activity_registered`.
4. Gunakan operator **`not in`** untuk memeriksa apakah aktivitas tersebut tidak terdapat dalam daftar.
5. Simpan hasilnya ke dalam variabel `activity_unknown`.

### Contoh 1

Jika mahasiswa memasukkan:

### Ketentuan

Gunakan **membership operators**:

```text

```

**Jangan menggunakan `if`, `else`, atau looping.**


````python
# ================================================
# STEP 7 - CAMPUS ACTIVITY CHECK
# ================================================

# List of campus activities available in the system
activities = ["KKN", "Magang", "Organisasi", "Praktikum Seminar TA"]

# Check whether the student's activity is registered
activity_registered = aktivitas in activities

# Check whether the student's activity is not registered
activity_unknown = aktivitas not in activities


# Display the results
print("\nActivity Check")
print("----------------------------------------")
# ================jawab disini================
print("Activity            :",aktivitas)
print("Registered Activity :",activity_registered)
print("Unknown Activity    :",activity_unknown)
````

# STEP 8 — SYSTEM STATUS

Pada tahap ini, sistem akan mengecek **status Campus Wallet** mahasiswa.

Ketika akun pertama kali dibuat, sistem **belum menentukan status akun**. Oleh karena itu, status disimpan dengan nilai:

```python
status = None
```

Nilai `None` berarti **belum ada nilai atau status yang ditentukan**.

Gunakan **Identity Operators** untuk mengecek kondisi tersebut.

### Tugas

1. Buat variable `status` dan isi dengan `None`.
2. Buat variable `status_none` untuk mengecek apakah `status` adalah `None`.

   * Gunakan operator `is`.
3. Buat variable `status_not_none` untuk mengecek apakah `status` bukan `None`.

   * Gunakan operator `is not`.
4. Tampilkan kedua hasil pemeriksaan.

### Expected Result

Karena pada awal program `status` masih bernilai `None`, maka:

```text
Identity Check
----------------------------------------
Status is None     : True
Status is not None : False
```

### Ketentuan

WAJIB menggunakan kedua **Identity Operators** berikut:

```text
is
is not
```

Jangan menggunakan `==` atau `!=` untuk pemeriksaan pada tahap ini.


````python
# ================================================
# STEP 8 - SYSTEM STATUS
# ================================================

# Status has not been determined yet
status = None


# Check whether status is None
status_none = status is None


# Check whether status is not None
status_not_none = status is not None


# Display the results
print("\nIdentity Check")
print("----------------------------------------")
# ================jawab disini================
print("Status is None     :",status_none)
print("Status is not None :",status_not_none)
````
